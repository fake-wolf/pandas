```
import scanpy as sc
import scvelo as scv
import matplotlib.pyplot as plt
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# 确保 numpy 版本符合要求
print(f"Current NumPy version: {np.__version__}")

# 1. 加载并预处理数据
adata = scv.datasets.pancreas()

# 运行速度分析流水线
scv.pp.filter_and_normalize(adata, min_shared_counts=20, n_top_genes=2000)
scv.pp.moments(adata, n_pcs=30, n_neighbors=30)
scv.tl.velocity(adata)
scv.tl.velocity_graph(adata)

# 2. 绘制矢量流线图 (High Fidelity)
fig, ax = plt.subplots(figsize=(9, 8), dpi=200)

scv.pl.velocity_embedding_stream(
    adata,
    basis='umap',
    color='clusters',
    legend_loc='right margin',
    density=1.3,
    smooth=0.7,
    arrow_size=1.3,
    arrow_color='#333333',
    linewidth=1.2,
    palette='viridis', # 或者使用 scVelo 默认配色
    ax=ax,
    show=False
)

# 3. 装饰
ax.set_title('RNA Velocity Streamflow (Refined)', fontsize=16, fontweight='bold', color='#444444')
ax.axis('off')

plt.tight_layout()
plt.show()
```

```
import scvelo as scv
import scanpy as sc
import matplotlib.pyplot as plt
import numpy as np

# 确认环境
print(f"NumPy version: {np.__version__}")

# 1. 加载数据
adata = scv.datasets.pancreas()

# 2. 修正后的预处理流程
# 使用 scanpy 进行核心预处理以确保兼容性
sc.pp.filter_genes(adata, min_cells=3)
scv.pp.filter_genes(adata, min_shared_counts=20)
scv.pp.normalize_per_cell(adata)
# 使用 scanpy 的高度变异基因筛选函数
sc.pp.highly_variable_genes(adata, n_top_genes=2000, flavor='seurat_v3' if 'counts' in adata.layers else 'cell_ranger')
sc.pp.log1p(adata)

# 计算矩、速度及图谱
scv.pp.moments(adata, n_pcs=30, n_neighbors=30)
scv.tl.velocity(adata)
scv.tl.velocity_graph(adata)

# 3. 绘制高保真流线图
fig, ax = plt.subplots(figsize=(8, 7), dpi=150)
scv.pl.velocity_embedding_stream(
    adata,
    basis='umap',
    color='clusters',
    density=1.2,
    smooth=0.7,
    arrow_size=1.2,
    ax=ax,
    show=False
)
ax.set_title('RNA Velocity Streamflow (Fixed)', fontsize=15, fontweight='bold')
plt.show()
```

```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.interpolate import griddata

# 1. 模拟生成城市 OD 数据 (起点 -> 终点)
np.random.seed(42)
n_trips = 1500

# 假设起终点在 [-3, 3] 的二维城市空间内
# 设置两个主要的吸引力中心 (例如 CBD 和 商业区)
cbd_center = np.array([1.5, 1.5])
suburb_center = np.array([-1.5, -1.5])

# 随机生成起点 (Origin)
o_x = np.random.uniform(-3, 3, n_trips)
o_y = np.random.uniform(-3, 3, n_trips)

# 根据引力模拟终点 (Destination)，大部分流向 CBD，少部分流向郊区中心
dest_choice = np.random.rand(n_trips) < 0.7
d_x = np.where(dest_choice,
               o_x + (cbd_center[0] - o_x) * 0.4 + np.random.normal(0, 0.3, n_trips),
               o_x + (suburb_center[0] - o_x) * 0.4 + np.random.normal(0, 0.3, n_trips))
d_y = np.where(dest_choice,
               o_y + (cbd_center[1] - o_y) * 0.4 + np.random.normal(0, 0.3, n_trips),
               o_y + (suburb_center[1] - o_y) * 0.4 + np.random.normal(0, 0.3, n_trips))

# 计算位移向量 (u, v) 与 移动距离 (Speed/Magnitude)
u = d_x - o_x
v = d_y - o_y
speed = np.sqrt(u**2 + v**2)

# 2. 将离散的 OD 点位插值转换为规则网格流场 (Grid Vector Field)
grid_x, grid_y = np.mgrid[-3:3:100j, -3:3:100j]

# 使用 scipy griddata 进行二维空间插值
grid_u = grid_data = griddata((o_x, o_y), u, (grid_x, grid_y), method='cubic', fill_value=0)
grid_v = griddata((o_x, o_y), v, (grid_x, grid_y), method='cubic', fill_value=0)
grid_speed = np.sqrt(grid_u**2 + grid_v**2)

# 3. 开始可视化绘图
fig, ax = plt.subplots(figsize=(9, 8), dpi=300)

# (1) 绘制底层移动强度热力图 (散点模拟网格背景)
background = ax.scatter(grid_x, grid_y, c=grid_speed, cmap='YlGnBu', s=8, alpha=0.7, edgecolors='none')

# (2) 绘制连续矢量流线 (Streamplot)
strm = ax.streamplot(
    grid_x[:, 0], grid_y[0, :], grid_u.T, grid_v.T,
    color='black',          # 流线颜色
    linewidth=1.2,          # 线宽
    density=1.4,            # 流线密度
    arrowsize=1.3,          # 箭头大小
    arrowstyle='->'         # 箭头样式
)

# (3) 标注主要城市功能区
ax.text(cbd_center[0], cbd_center[1], 'CBD (Main Sink)', fontsize=11, fontweight='bold',
        color='darkred', ha='center', bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.8))
ax.text(suburb_center[0], suburb_center[1], 'Sub-Center', fontsize=10, fontweight='bold',
        color='darkblue', ha='center', bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.8))

# 4. 图面美化
ax.set_title('Urban Mobility Vector Field (OD Streamplot)', fontsize=14, color='#2c3e50', fontweight='bold', pad=15)
ax.axis('off')

# 添加底图颜色条
cbar = fig.colorbar(background, ax=ax, shrink=0.7, pad=0.03)
cbar.set_label('Flow Intensity / Speed', fontsize=10)

plt.tight_layout()
plt.show()
```

The Python SDK for the Gemini API is contained in the [`google-generativeai`](https://pypi.org/project/google-generativeai/) package. Install the dependency using pip:

To use the Gemini API, you'll need an API key. If you don't already have one, create a key in Google AI Studio.

Before you can make any API calls, you need to initialize the Generative Model.

```
model = genai.GenerativeModel('gemini-pro')
```

## Generate text
